```python
# Baixando o arquivo de demografia
# URL do IBGE
#    ↓
# requests.get()
#    ↓
# conteúdo do ZIP
#    ↓
# memória do notebook
response_demografia = requests.get(url_demografia)

response_demografia.raise_for_status() # Verifica se deu certo (200)

arquivo_zip = response_demografia.content

print(f"Tamanho do arquivo: {len(arquivo_zip) / (1024 * 1024):.2f} MB")
```

```python
# biblioteca para lidar com arquivos ZIP nativa do Python
import zipfile
import io

zip_ibge = zipfile.ZipFile(io.BytesIO(arquivo_zip))

zip_ibge.namelist()
```

```python
# Ler o arquivo zip com pandas
df_demografia = pd.read_csv(
    zip_ibge.open("Agregados_por_bairros_demografia_BR.csv"),
    sep=";",
    encoding="latin1"
)

df_demografia.head()
# df_demografia.shape
# df_demografia.columns.tolist()

# No próprio dicionário do IBGE a coluna V01006 é a de "Quantidade de moradoes"
```

```python
df_chamados = pd.DataFrame(chamados)

df_chamados
```

```python
# Contar quantos chamados existem em cada bairro

df_chamados_por_bairro = (
    df_chamados
    .groupby("bairro")
    .size()
    .reset_index(name="chamados")
)

df_chamados_por_bairro
```

```python
df_populacao_recife = df_recife[
    ["CD_BAIRRO", "NM_BAIRRO", "V01006"]
].copy()

df_populacao_recife
```

```python
df_populacao_recife = df_populacao_recife.rename(
    columns={
        "CD_BAIRRO": "codigo_bairro",
        "NM_BAIRRO": "bairro",
        "V01006": "populacao"
    }
)

df_populacao_recife
```

```python
# Padronizando os nomes dos bairros para facilitar a junção dos DataFrames

df_chamados_por_bairro["bairro"] = (
    df_chamados_por_bairro["bairro"]
    .str.strip()
    .str.upper()
)

df_populacao_recife["bairro"] = (
    df_populacao_recife["bairro"]
    .str.strip()
    .str.upper()
)
```

```python
# Merge dos DataFrames para unir os dados de chamados e população por bairro

df_bairros = df_chamados_por_bairro.merge(
    df_populacao_recife[["codigo_bairro", "bairro", "populacao"]],
    on="bairro",
    how="left"
)

df_bairros
```

```python
df_bairros["populacao"] = pd.to_numeric(
    df_bairros["populacao"],
    errors="coerce"
)
```

```python
df_bairros["chamados_por_1000"] = (
    df_bairros["chamados"] / df_bairros["populacao"]
) * 1000


df_bairros["chamados_por_1000"] = (
    df_bairros["chamados_por_1000"]
    .round(2)
)

#"Boa Viagem teve X chamados para cada 1.000 habitantes.!"
# Isso torna a comparação entre bairros muito mais justa, porque um bairro grande 
# naturalmente tende a ter mais chamados simplesmente por ter mais moradores.
```

```python
# Etapa de Load!

import os
from dotenv import load_dotenv
from pymongo import MongoClient
from pymongo.server_api import ServerApi

load_dotenv()

mongo_uri = os.getenv("MONGODB_URI")

client = MongoClient(
    mongo_uri,
    server_api=ServerApi("1")
)

client.admin.command("ping")

print("Conexão com MongoDB realizada com sucesso!")
```

```python
# Vamos guardar o dataFrame da população total do IBGE!
# Criando banco da coleção e inserindo os dados do DataFrame no MongoDB

db = client["FISCALIZE_ETL"]

collection = db["populacao_ibge_raw"]

dados_brutos = df_demografia.to_dict(orient="records")
resultado = collection.insert_many(dados_brutos)

print(f"Registros inseridos: {len(resultado.inserted_ids)}")
```

```python
# Vamos adicionar no SQLite coleção de chamados por bairro,
# com a população e a taxa de chamados por 1.000 habitantes.

import sqlite3

conn = sqlite3.connect("fiscalize_etl.db")

df_bairros.to_sql(
    "chamados_populacao",
    conn,
    if_exists="replace",
    index=False
)

conn.close()
```
