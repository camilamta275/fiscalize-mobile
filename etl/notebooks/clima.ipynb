```python
df_chamados = pd.DataFrame(chamados)

df_chamados
```

```python
url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": -8.0631,
    "longitude": -34.8711,
    "start_date": "2026-02-01",
    "end_date": "2026-02-28",
    "daily": "temperature_2m_mean,precipitation_sum",
    "timezone": "America/Recife"
}

response = requests.get(url, params=params)

response.raise_for_status() # Verifica se a requisição foi bem-sucedida, caso contrário, lança uma exceção.

clima = response.json() # Armazena a resposta da API em formato JSON na variável 'clima'.

clima
```

```python
# Cria um DataFrame a partir dos dados de clima obtidos da API, contendo as colunas "data", "precipitacao" e "temperatura".
df_clima = pd.DataFrame({
    "data": clima["daily"]["time"],
    "precipitacao": clima["daily"]["precipitation_sum"],
    "temperatura": clima["daily"]["temperature_2m_mean"]
})

df_clima
```

```python
df_chamados["data"] = pd.to_datetime(df_chamados["criadoem"]).dt.date # Converte a coluna "criadoem" do dataFeame de Chamados para o tipo datetime e extrai apenas a data, removendo a parte de tempo.

df_clima["data"] = pd.to_datetime(df_clima["data"]).dt.date # Converte a coluna "data" do dataFrame de Clima para o tipo datetime e extrai apenas a data, removendo a parte de tempo.

# df_chamados[["id", "data"]] # Exibe apenas as colunas "id" e "data" do DataFrame de Chamados.
df_clima.head() # Exibe as 5 primeiras linhas do DataFrame de Clima.
```

```python
# Mescla os DataFrames de Chamados e Clima com base na coluna "data", 
# utilizando um merge do tipo "left" para manter todas as linhas do DataFrame de Chamados e adicionar as informações de clima correspondentes.
#Quando houver uma data correspondente em df_clima, os valores de precipitacao e temperatura
df_final = df_chamados.merge(
    df_clima,
    on="data",
    how="left"
)

df_final # Crição do dataFrame final que contém as informações de chamados e clima mescladas com base na coluna "data".
```

```python
# Extract:
# 1. JSON fictício
# 2. API pública Open-Meteo

# Transform:
# 1. JSON → DataFrame
# 2. padronização de datas
# 3. seleção/renomeação de campos
# 4. JOIN entre chamados e clima
# 5. criação do DataFrame final

# Load:
# df_final
#    │
#    ├── MongoDB → dados brutos
#    │
#    └── SQLite  → dados transformados
```

```python
# Se conectando ao MongoDB para salvar o dataFrame Clima
import os
from dotenv import load_dotenv
from pymongo import MongoClient
from pymongo.server_api import ServerApi

load_dotenv()

mongo_uri = os.getenv("MONGODB_URI")

client = MongoClient(
    mongo_uri,
    server_api=ServerApi("1")
)
```

```python
client.admin.command("ping") # testando conexão com o MongoDB
```

```python
# Criando um banco específico para armazenar o dado bruto do dataFrame Clima.
db = client["FISCALIZE_ETL"]

collection = db["clima_raw"]

collection.insert_one(clima)
```

```python
# Testando pra ver se realmente salvou o dado no MongoDB, buscando o primeiro documento da coleção "clima_raw".
documento = collection.find_one()

import pprint

pprint.pp(documento)
```

```python
import sqlite3
conn = sqlite3.connect("fiscalize_etl.db")

# Salvando o DataFrame final no banco de dados SQLite, na tabela "chamados_clima".
df_final.to_sql(
    "chamados_clima",
    conn,
    if_exists="replace",
    index=False
)

conn.close()
```

```python
#Conferindo o SQLite, abrindo uma conexão e lendo os dados da tabela "chamados_clima" para um DataFrame.
conn = sqlite3.connect("fiscalize_etl.db")

df_teste = pd.read_sql(
    "SELECT * FROM chamados_clima",
    conn
)

conn.close()

df_teste
```
